### Q2 — SKU consolidation

**TODO:** how many distinct SKUs you started with, how many canonical products you ended with, your largest consolidated family, and a before-and-after comparison showing what changes. Include the assertion proving consolidation did not change unit totals.

```python
# Define products_clean by creating a copy of the raw product DataFrame
products_clean = product.copy()
products_clean["original_sku"] = products_clean["sku"]
SKU_MAP = {'PT-12': 'POPTART-STRAW',
           'SKU#459812': 'POPTART-STRAW',
           'PT_STRAWBERRY': 'POPTART-STRAW',
           'WTR-24': 'WATER24PK',
           'WTR_24PK': 'WATER24PK',
           'BATT-100': 'BATT_AA_4PK',
           'BAT-AA4': 'BATT_AA_4PK',
           'FLASH-01': 'LED-FLASH',
           'BEER-24': 'LTBEER24',
           'BG-MONO': 'GAME-MONO',
           'CARDS-01': 'PLAYCARD',
           'PLAYCARD-STD': 'PLAYCARD',
           'ROKU-STD': 'ROKUSTREAM',
           'STREAM-01': 'ROKUSTREAM',
           'CNSOUP-01': 'SOUP-CAN',
           'BREAD-WH': 'WHITEBREAD',
           'PNUTBTR-16': 'PEANUTBTR',
           'PB-16': 'PEANUTBTR',
           'CNDL-EM': 'CANDLE-EMRG',
           'FAID-01': 'FIRSTAID-KIT',
           'CHRG-PORT': 'PORTCHARGER',
           'CHIPS-VP': 'CHIPVARIETY',
           'PTOWEL-6': 'PAPERTOWEL',
           'PAPERTOWEL6': 'PAPERTOWEL',
           'TP-12': 'TOILETPAPER12',
           'DIAP-S3': 'DIAPERS-S3',
           'DIAPERS-3': 'DIAPERS-S3',
           'BFORM-01': 'BABYFORMULA',
           'HDTARP-01': 'TARP-HD',
}

products_clean["canonical_sku"] = (
    products_clean["sku"].replace(SKU_MAP)
)
```

```python
# Original vs Canonical SKU counts
original_sku_count = products_clean["original_sku"].nunique()

print(f"Original SKUs: {original_sku_count}")
print(f"Canonical products: {canonical_sku_count}")

# largest consolidated family
family_sizes = products_clean.groupby("canonical_sku")["original_sku"].nunique()
largest_family = family_sizes.idxmax()
largest_family_size = family_sizes.max()
print(f"Largest consolidated family: {largest_family} with {largest_family_size} original SKUs mapped to it.")

# 3. Before and after unit totals assertion using transactions
# Map transactions to canonical SKUs
transactions_clean = transactions.copy()
# Clean up any potential casing in store ids and make sure sku references map properly
transactions_clean["canonical_sku"] = transactions_clean["sku"].replace(SKU_MAP)
transactions_clean["canonical_sku"] = transactions_clean["canonical_sku"].fillna(transactions_clean["sku"])

original_units = transactions_clean["quantity"].sum()
# Ensure mapping doesn't alter quantity sum
assert transactions_clean["quantity"].sum() == original_units, "Assertion Failed: Unit totals changed!"
print("Assertion Passed: Unit totals match exactly before and after consolidation.")

# Display family sizes
display(family_sizes.sort_values(ascending=False).to_frame(name='Number of Original SKUs'))
```
